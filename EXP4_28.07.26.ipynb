```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer,SimpleImputer
df=pd.DataFrame({'Car Model':[2022,2025,2023,np.nan,2022],'Price':[1000000,10000000,2500000,1800000,np.nan],'Engine capacity(in cc)':[np.nan,5900,2500,1600,800]})
df['Car Model']=df['Car Model'].astype('Int64')
df
```

```python
mean_imputer=SimpleImputer(strategy="mean")
df["Price_mean"]=mean_imputer.fit_transform(df[["Price"]])
df
```

```python
mode_imputer=SimpleImputer(strategy="most_frequent")
df["Car_model_mode"]=mode_imputer.fit_transform(df[["Car Model"]])
df
```

```python
median_imputer=SimpleImputer(strategy="median")
df["Engine_capacity_median"]=median_imputer.fit_transform(df[["Engine capacity(in cc)"]])
df
```

```python
num_cols = ["Price", "Engine capacity(in cc)"]
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice = df[num_cols].copy()
df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)
df["Price_MICE"] = df_mice["Price"]
df["Engine Capacity(in cc)_MICE"] = df_mice["Engine capacity(in cc)"]
print(df) 
```
