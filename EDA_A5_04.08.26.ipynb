```python
#Dataset with outliers
from sklearn.preprocessing import MinMaxScaler,StandardScaler,RobustScaler
data={'Birthyear':[2000,2005,2004,2007,2001],'Income':[50000,20000,30000,60000,300900]}
df2=pd.DataFrame(data)
print("Original data\n",df2,"\n")
```

```python
#Min-Max scaling
mm_scaler=MinMaxScaler()
df_mm=pd.DataFrame(mm_scaler.fit_transform(df2),columns=df2.columns)
print("Min-Max Scaled Data\n",df_mm,"\n")
```

```python
#Z-Score Standardization
ss=StandardScaler()
df_ss=pd.DataFrame(ss.fit_transform(df2),columns=df2.columns)
print("Z-Score Standardized Data\n",df_ss,"\n")
```
