```python
import pandas as pd
import numpy as np
df=pd.read_csv("F:\EDA\drug200.csv")
print(df)
```

```python
print(df.tail(10))
```

```python
df1=pd.DataFrame({'Age':[45,34,67],'Sex':['M','F','M'],'BP':['HIGH','LOW','HIGH'],'Cholestrol':['HIGH','LOW','LOW'],'Na_to_K':[12.3,23.66,67.34],'Drug':['DrugB','DrugC','DrugY']})
vertical_concat=pd.concat([df,df1],ignore_index=True)
print("Vertical concatination")
print(vertical_concat)
```

```python
df2=pd.DataFrame({
    'Name':[
'ram','sam','gokul','ramesh','suresh','deepak','lexmi','sathish','yamal','messi']
})
horizontal_concat=pd.concat([df,df2],axis=1)
print("\nHorizontal Concatination:")
print(horizontal_concat)
```

```python
df3=pd.DataFrame({'Age':[23,34,56,67,89],'Health score':[45,68,78,77,67]})
inner_joint=pd.merge(df,df3,on='Age',how='inner')
print(inner_joint)
```

```python
left_joint=pd.merge(df,df3,on='Age',how='left')
print(left_joint)
```

```python
right_joint=pd.merge(df,df3,on='Age',how='right')
print(right_joint)
```

```python
outer_joint=pd.merge(df,df3,on='Age',how='outer')
print(outer_joint)
```

```python
Drug_price=pd.DataFrame({'Quantity':[10,20,22,11,5,6],'Price_per_item':[200,300,300,60,80,90]})
Drug_price['Totalprice']=Drug_price['Quantity']*Drug_price['Price_per_item']
print("\nDrug price data")
print(Drug_price)
```

```python
s1=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
s2=pd.Series([6,7,8],index=['f','g','h'])
f=pd.concat([s1,s2],keys=['N1','N2'])
print("\n Hierarchical Series:")
print(f)
print("\nUnstacked Series")
print(f.unstack())
```
